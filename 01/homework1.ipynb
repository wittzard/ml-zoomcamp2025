```python
#A7
df_filter = df[df['origin'] == 'Asia']
df_filter = df_filter[['vehicle_weight','model_year']]
df_filter = df_filter.head(7)
```

```python
X = df_filter.to_numpy()
X_T = X.T
XTX = np.dot(X_T, X)
XTX_inv = np.linalg.inv(XTX)

y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])

w = (XTX_inv @ X_T) @ y
w.sum()
```
